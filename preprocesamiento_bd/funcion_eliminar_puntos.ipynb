```python
import pandas as pd
import numpy as np
```

```python
crop_rec = pd.read_csv('crop_recommendation_unida.csv')
crop_rec.head()
```

```python
ensayo = crop_rec.copy()
```

```python
def add_char_index(cadena, char = '.', i = 0, remove_char = False):
    if remove_char == True:
        cadena = str(cadena.replace(char, ''))
    left = cadena[:i]
    right = cadena[i:]
    return float(left + char + right)
```

```python
ensayo['temperature'] = ensayo['temperature'].apply(lambda x: add_char_index(x, '.', 2, True))
ensayo['humidity'] = ensayo['humidity'].apply(lambda x: add_char_index(x, '.', 2, True))
ensayo['ph'] = ensayo['ph'].apply(lambda x: add_char_index(x, '.', 1, True))
ensayo['rainfall'] = ensayo['rainfall'].apply(lambda x: add_char_index(x, '.', 3, True))

ensayo.head()
```

```python
columnas = ['temperature', 'humidity', 'ph', 'rainfall']
ensayo[columnas] = ensayo[columnas].round(3)
ensayo.head()
```

```python
ensayo.to_excel('crop_recommendation_corregida.xlsx')
```
